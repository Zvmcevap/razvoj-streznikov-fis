```python
import pika
import time
import numpy as np

credentials = pika.PlainCredentials('martin', 'martin00')
parameters =  pika.ConnectionParameters('149.62.71.186', credentials=credentials)
connection = pika.BlockingConnection(parameters)
channel = connection.channel()

#Queue ostane živ tudi po restartu RabbitMQ (ampak zgolj metadata!)
channel.queue_declare(queue='task', durable=True)

#MQrabbit zagotovi, da bo sporočilo zapisano! ===> PERSISTENT, isto kot delivery_mode = 2
properties = pika.BasicProperties(delivery_mode=pika.spec.PERSISTENT_DELIVERY_MODE)

for i in range(4):
    body=str(np.random.randint(1,3))
    print(f"task {body} given away!")
    channel.basic_publish(exchange='', routing_key='task', body=body, properties=properties)
    
```

The provided Python code snippet successfully simulates sending tasks to a message queue. Here's a more concise explanation:

**Libraries:**

* `pika`: Connects to and interacts with message brokers using AMQP.
* `time` (unused here): Could be used to introduce delays between tasks.
* `numpy`: Generates random integers for the message content.

**Connection and Channel:**

* Establishes a connection to the message broker at "149.62.71.186" with username "martin" and password "martin00".
* Creates a channel on that connection for communication.

**Queue Setup:**

* Declares a durable queue named "task". Durable queues persist even if the broker restarts.

**Publishing Tasks:**

* Defines message properties to make messages persistent (not lost on broker crashes).
* A loop iterates four times:
    * Generates a random integer between 1 and 2 (inclusive) and converts it to a string as the message body.
    * Prints a message indicating a task has been created.
    * Publishes the message to the "task" queue with the empty exchange (''), sending it directly to the queue.

**In summary, this code sends four messages with random integers (1 or 2) to a persistent queue named "task" on the message broker.** 

**Additional Notes:**

* Remember to replace "149.62.71.186" with the actual IP address of your message broker.
* This code uses synchronous communication with `BlockingConnection`. For handling high message volume, consider asynchronous approaches.
