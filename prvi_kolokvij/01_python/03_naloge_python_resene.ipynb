## naloga 5

Napišite dekorator, ki vsak rezultat funkcije (tisto, kar vrne) zapišem v neko lokalno text datoteko in tudi sprintamo uporabniku.

```python
import time
def log_function_return(func):
    print("starting!")
    def wrapper(*args, **kwargs):
        res = func(*args, **kwargs)
        with open("log_func.txt", "a", encoding="utf-8") as datoteka:
            datoteka.write(str(res)+"\n")
        return res
    return wrapper




@log_function_return
def calc_sum(a,b):
    return a+b

res = calc_sum(1,5)
print(res)
```

## Naloga 6

V JSON obliki (shranjeno v datoteki) shranjuj in vodi imenik (ime, oddelek, datum zaposlitve) svojih zaposlenih. Naredi 2 funkciji: odpusti (dobi ime in ga izbriši) in zaposli. 

```python
import json
from datetime import datetime

#datum = datetime(2020, 5, 25) # leto, mesec, dan, ura, minuta

ekipa = [
    {"ime": "Luka", "oddelek": "IT", "datum_zaposlitve": datetime(2020, 5, 25).isoformat()},
    {"ime": "Maja", "oddelek": "marketing", "datum_zaposlitve": datetime(2009, 1, 21).isoformat()}
]

# Save to a .json file
with open('imenik.json', 'w') as file:
    json.dump(ekipa, file, indent=1)

nov_clan = {"ime": "Ernest", "oddelek": "son of the boss", "datum_zaposlitve": datetime.now().isoformat()}

def dodaj_clana(clan):
    with open('imenik.json', 'r') as file:
        data = json.load(file)
    data.append(clan)
    
    with open('imenik.json', 'w') as file:
        json.dump(data, file, indent=1)

dodaj_clana(nov_clan)
```

```python
def odpusti_clana(ime_clana):
    with open('imenik.json', 'r') as file:
        data = json.load(file)
    index = -1
    for i in range(0,len(data)):
        if data[i]["ime"]==ime_clana:
            index = i
    if(index==-1):
        print("error, member not found")
        return 
    else:
        print(f"{ime_clana}, you are fired!")
        data.pop(index)
    
    with open('imenik.json', 'w') as file:
        json.dump(data, file, indent=1)
odpusti_clana("Ernest")
```
