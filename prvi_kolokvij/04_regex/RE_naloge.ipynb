### Regex Explanation

```python
p = re.compile(r'\b(?:(?<=and)\s|^|\.\s)([a-z]+)', re.IGNORECASE)
```

This regular expression is designed to **match lowercase words that appear at specific positions in a sentence**, with case-insensitive matching enabled.

#### Breakdown

* `\b`
  Matches a **word boundary**, ensuring the match starts at the beginning of a word.

* `(?: ... )`
  A **non-capturing group** that defines the allowed contexts before the word:

  * `(?<=and)\s`
    A **positive lookbehind** that matches a space only if it is preceded by `"and"`
    → captures words that come immediately after `"and"`

  * `^`
    Matches the **start of the string**

  * `\.\s`
    Matches a **period followed by a space**
    → captures words at the start of a new sentence

* `([a-z]+)`
  A **capturing group** that matches one or more alphabetic characters (`a–z`)
  → this is the actual word being extracted

* `re.IGNORECASE`
  Makes the pattern **case-insensitive**, so `[a-z]+` also matches uppercase letters

#### What it Matches

This pattern extracts words that:

* Appear at the **beginning of the text**
* Appear **right after a period and space** (start of a sentence)
* Appear **immediately after the word "and"**

---


````python
p = re.compile(r'\b(?:(?<=and)\s|^|\.\s)([a-z]+)', re.IGNORECASE)
m = p.findall('Buying chesse AND salami AND wine AND bread. Soup AND veggies.')
print(m)
````


```python
p = re.compile(r'(^\w+\b)|and\s*(\b\w+\b)|[.]\s+(\b\w+\b)', re.IGNORECASE)
```

This pattern matches **words that occur in three specific contexts**: at the start of the string, after the word `"and"`, or after a period.

#### Breakdown

The regex uses **alternation (`|`)** to define three separate matching cases:

---

**1. `(^\w+\b)`**

* `^` → start of the string
* `\w+` → one or more word characters (letters, digits, underscore)
* `\b` → word boundary

Matches the **first word in the string**
---

**2. `and\s*(\b\w+\b)`**

* `and` → literal match
* `\s*` → optional whitespace
* `(\b\w+\b)` → captures a full word

Matches a **word that follows "and"**
---

**3. `[.]\s+(\b\w+\b)`**

* `[.]` → literal period (`.`)
* `\s+` → one or more spaces
* `(\b\w+\b)` → captures a full word

Matches a **word at the start of a sentence after a period**
---

#### Groups

* Group 1 → word at the start of the string
* Group 2 → word after `"and"`
* Group 3 → word after a period

Only **one group is populated per match**, depending on which alternative matched.

---

#### Flags

* `re.IGNORECASE`
  Makes matching **case-insensitive**, so `"And"` or `"AND"` also work.

---

#### Summary

This regex extracts words that appear:

* At the **beginning of the text**
* Immediately after **"and"**
* Immediately after a **sentence-ending period**

Each case is handled independently via alternation, rather than shared lookarounds as in the previous pattern.


````python
#alternativna resitev
p = re.compile(r'(^\w+\b)|and\s*(\b\w+\b)|[.]\s+(\b\w+\b)', re.IGNORECASE)
m = p.findall('Buying chesse AND salami AND wine AND bread. Soup AND veggies.')

print(m)
````

2. naloga


* `|` (alternation)
  Splits the regex into **three independent patterns**. Unlike the previous version (which used a shared context), each branch here works separately.

* `\w` instead of `[a-z]`
  Matches **any word character**:

  * letters (`a–z`, `A–Z`)
  * digits (`0–9`)
  * underscore (`_`)
    → broader than `[a-z]+`

* `\b` at the *end* of `\w+\b`
  Ensures the match **stops at a word boundary**, preventing partial matches inside longer tokens.

* `\s*` vs `\s`

  * `\s*` → **zero or more** spaces (more permissive)
  * Previously you required exactly one space via `\s`

* `\s+` after `.`
  Requires **at least one space** after a period (stricter than just matching `". "` literally)

* `[.]` instead of `\.`
  Functionally identical here (both match a literal `.`), but `[.]` uses a **character class**, which is unnecessary but valid.

* Multiple capturing groups `( ... )` across alternatives
  Each branch has its **own capture group**, so:

  * Only one group is populated per match
  * You must check which group is non-`None`

````python
p = re.compile(r'\b([aeiou][a-z]*[^aeiou.])(?:\s|$|[.])', re.IGNORECASE)
m = p.findall('ABC ded aba UEB AOB. aaa ded aba UEB aoa.')
print(m)
````

````python
#alternativna resitev
p = re.compile(r'\b[aeiou]\w+[^aeiou\s]\b', re.IGNORECASE)
m = p.findall(r'ABC ded aba UEB AOB, aaa ded aba UEB aoa.')
print(m)
````

````python
p = re.compile(r'(?<=\().*?(?=\))', re.IGNORECASE)
m = p.findall('Tezka naloga (3.naloga) je tezka ampak ni tako tezka kot (prva) naloga.')
#m = p.findall('abc')
print(m)
````

````python
#alternativa resitev
p = re.compile(r'(?<=[(])[\w\d.\s]+(?=[)])', re.IGNORECASE)
m = p.findall('Tezka naloga (3.naloga) tezka ampak ni tako tezka kot (prva) naloga.')
print(m)
````
