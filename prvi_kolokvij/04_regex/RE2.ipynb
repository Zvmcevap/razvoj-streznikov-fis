## Regularni izrazi

Regularni izrazi(regex) poganjajo številne vidike računalništva, vključno s prevajalniki. Na področju prevajalnikov imajo redni izrazi ključno vlogo pri prevajanju kode v strojno berljiva navodila. 

**Jedrna funkcionalnost:**

* **Iskanje vzorcev:** Redni izrazi omogočajo določanje vzorcev s kombinacijo dobesednih znakov, posebnih znakov in kvantifikatorjev. Ti vzorci se lahko ujemajo z določenimi zaporedji, znaki ali razponi znotraj besedila.
* **Iskanje in pridobivanje:** Na podlagi definiranega vzorca lahko redni izrazi učinkovito preiščejo podatke besedila in najdejo vse pojavitve, ki ustrezajo merilom. To omogoča ekstrakcijo specifičnih informacij iz večjih naborov podatkov.

**Zgodovinski kontekst:**

Koncept rednih izrazov izvira iz 50. let prejšnjega stoletja na področju matematike, kjer so jih uporabljali za opisovanje vzorcev v jezikih. Njihova uporabnost za obdelavo besedil v računalnikih je bila hitro prepoznana, kar je vodilo k njihovi integraciji v različne programske jezike.

**Prednosti rednih izrazov:**

* **Učinkovitost:** Redni izrazi odlično obvladajo hitro obdelavo velikih količin podatkov besedila. Avtomatizirajo lahko kompleksne preiskave, ki bi bile pri ročnem izvajanju zamudne ali nagnjene k napakam.
* **Natančnost:** S natančnim določanjem vzorcev iskanja redni izrazi zagotavljajo, da se prepozna samo relevantno besedilo, kar zmanjša tveganje napak pri ročnem iskanju.
* **Prožnost:** Redni izrazi ponujajo bogato sintakso za določanje vzorcev iskanja. Lahko obvladajo široko paleto meril iskanja, od preprostih zaporednih znakov do zapletenih kombinacij črk, številk in simbolov.

**Uporaba:**

* **Prevajalniki:** Redni izrazi igrajo ključno vlogo v prevajalnikih med leksikalno analizo (pregledovanjem) in sintaksno analizo (sintaksnim razčlenjevanjem). Pomagajo identificirati pomene besede (kot so ključne besede ali identifikatorji) in zagotoviti, da se koda drži slovničnih pravil jezika.
* **Iskalniki:** Redni izrazi poganjajo iskalnike, saj jim omogočajo razumevanje uporabniških poizvedb in identifikacijo spletnih strani, ki ustrezajo namenu iskanja.
* **Validacija podatkov:** Redni izrazi se uporabljajo za validacijo uporabniškega vnosa v obrazcih, s čimer se zagotovi skladnost s specifičnimi formati (npr. e-poštni naslovi, telefonske številke).
* **Obdelava besedil:** Redni izrazi se uporabljajo v različnih orodjih za obdelavo besedil za napredno iskanje in zamenjavo, pridobivanje podatkov in manipulacijo z besedilom.

**Učenje rednih izrazov:**

Čeprav se redni izrazi na prvi pogled morda zdijo zapleteni, ponujajo dragoceno znanje za programerje, vse, ki delajo s podatki besedila, in tiste, ki jih zanima razumevanje prevajalnikov. S predano prakso in raziskovanjem lahko pridobite znanje o uporabi rednih izrazov za racionalizacijo opravil obdelave besedil, odklepanje novih možnosti manipulacije podatkov in poglobljeno razumevanje sveta prevajalnikov.


```python
import re
import sys
print(sys.version)
```

metacharacters   . ^ $ * + ? {m,n } [] \ | ( )

```python
p = re.compile('\\\\next')
m = p.match('\\next')
print(m)
```

```python
p = re.compile(r'\\next')
m = p.match(r'\next')
print(m)
```

```python
p = re.compile(r'.*abc')
m = p.match(r'abcdadadsdasd')
print(m)
```

```python
print(m.group(),m.start(), m.end(),m.span())
```

```python
p = re.compile('ab*', re.IGNORECASE)
m = p.match('AB')
print(m)
```

```python
p = re.compile(r'\bclass\b')
print(p.search('no class at all'))
print(p.search('the declassified algorithm'))
print(p.search('one subclass is'))
```

```python
s = '<html><head><title>Title</title>'
print(re.match('<.*>', s))
print(re.match('<.*?>', s))
```

```python
p = re.compile('(a(b)c)d')
m = p.match('abcd')
print(m.group(0))
print(m.group(1))
print(m.group(2))
```

```python
p = re.compile(r'\b(\w+)\s+\1\b')
p.search('Paris in the the spring').group()
```

```python
p = re.compile(r'(?P<word> hello) world')
m = p.search( 'hello hello world' )
print(m.group('word'))
print(m.group(1))
print(m.group(0))
```

```python
p = re.compile(r'\b(?P<word>\w+)\s+(?P=word)\b')
p.search('Paris in the the spring').group()
```

```python
m = re.match("([abc])+", "abc")
print(m.groups())
print(m.group(0))
```

```python
m = re.match("(?:[abc])+", "abc")
print(m.groups())
print(m.group(0))
```

```python
m = re.match("([abc]+)", "abc")
print(m.groups())
print(m.group(0))
```

Negative lookahead assertion. This is the opposite of the positive assertion; it succeeds if the contained expression doesn’t match at the current position in the string.

```python
p = re.compile(r'.*[.](?!bat$|exe$)([^.]*)$')
m = p.search('ah.bat.exe.wow')
if m:
    print(m.groups())
else:
    print('Empty!')
```

```python
p = re.compile(r'(?>krokodil|krokodilck)i')
m = p.search('krokodili')
if m:
    print(m.group())
else:
    print('Empty!')
```

```python
text = "This is a test string with HTML tags <html> and <body>"
pattern = r"<.*?>"  # Matches any characters between '<' and '>' (greedy)

matches = re.findall(pattern, text)
print(matches)
```

```python
text = "This is a test string with HTML tags <html<body>>"
pattern = r"<.*?>"  # Matches any characters between '<' and '>' (greedy)

matches = re.findall(pattern, text)
print(matches)
```
