## try/except

```python
import traceback

def age_ratio(age1, age2):
    if age1 < 0 or  age2 < 0 :
        raise ValueError("Age cannot be negative.")
    print(f"Ratio is: {age1/age2}")

try:
    user_age_1 = int(input("Enter your 1st age: "))
    user_age_2 = int(input("Enter your 2st age: "))
    age_ratio(user_age_1, user_age_2)
except ValueError as e:
    print("Error:", e)
    traceback.print_exc()
except Exception as e:
    print(f"An unexpected error occurred: {e}")
    print(f"Exception type: {type(e)}")
    traceback.print_exc()
print('Program has ended.')
```

## Mutable/non mutable

```python
# Mutable  (list)
my_list = [1, 2, 3]
another_list = my_list
another_list.append(4)
print(my_list)  

# Immutable  (integer)
x = 10
y = x
y = 20
print(x) 
```

```python
class MutableOrImmutable:
    def __init__(self, value, mutable=True):
        self._value = value
        self._mutable = mutable

    @property
    def value(self):
        return self._value

    @value.setter
    def value(self, new_value):
        if self._mutable:
            self._value = new_value
        else:
            raise TypeError("Object is immutable")
    

    def __repr__(self):
        return f"(value={self._value}, mutable={self._mutable})"

# Example usage:

# Mutable object
mutable_obj = MutableOrImmutable([1, 2, 3])
print(mutable_obj)
mutable_obj.value.append(4)  # Modifies the list in place
print(mutable_obj)
mutable_obj.value = [5,6,7] #reassigns the list
print(mutable_obj)
```

```python
#attempting to modify a list that is set to immutable.
immutable_list = MutableOrImmutable([1,2,3], mutable=False)
print(immutable_list)

#attempting to reassign a new list to the immutable object
try:
  immutable_list.value.append(4)  # Modifies the list in place
except TypeError as e:
  print(e)
print(immutable_list)
```

```python
import time
def log_execution_time(func):
    print("starting!")
    def wrapper(*args, **kwargs):
        current_time = time.time()
        func(*args, **kwargs)
        print(f"finished execution  in {time.time()-current_time}")
    return wrapper




@log_execution_time
def fibbonaci(n):
    a = 0
    b = 1
    ne = b  
    count = 1

    while count <= n:
        print(ne, end=" ")
        count += 1
        a, b = b, ne
        ne = a + b
x=10
fibbonaci(x)
```

## garbage collection

```python
# Garbage collection
def create_objects():
    imut_obj1 = 1
    imut_obj2 = imut_obj1
    imut_obj2 = imut_obj2*2
    
    mut_obj1 = [1, 2, 3]  
    mut_obj2 = mut_obj1 
    #what out! there's difference between in-place modifications and reassignment!
    mut_obj2.append(4)
    #mut_obj2 = mut_obj2*2
    print(f"Mutable object created: {id(mut_obj1)},  {id(mut_obj2)}")
    print(f"Immutable object created: {id(imut_obj1)},  {id(imut_obj2)}")
    # All these objects will become unreachable after the function returns.

create_objects()
# At this point, obj1 and obj2 are no longer referenced, and the garbage collector will eventually reclaim their memory.

#print(imut_obj1)
import gc
print(gc.get_threshold()) #prints the garbage collection thresholds.

# force garbage collection
#This function explicitly forces a full garbage collection cycle.
#It returns the number of unreachable objects found.
gc.collect()
```

```python
import gc

def create_cycle():
    list_a = []
    list_b = []
    list_a.append(list_b) # a points to b
    list_b.append(list_a) # b points to a


create_cycle()
print(gc.collect()) 
```

```python
class Car:
  def __init__(self, brand, model, color):
    self.brand = brand
    self.model = model
    self.color = color

  def accelerate(self):
    print(f"The {self.brand} {self.model} engine is fuming!")
```

```python
class ElectricCar(Car):
  def __init__(self, brand, model, color, battery_type):
    super().__init__(brand, model, color)  # Call parent constructor
    self.battery_type = battery_type

  # Override the accelerate function for electric cars
  def accelerate(self):
    print(f"The {self.brand} {self.model} is accelerating silently!")
```

```python
avto1 = Car('BMW', '220i', 'pink')
avto2 = ElectricCar('Tesla', 'S', 'rose', 'AAA')

avto1.accelerate()
avto2.accelerate()
```

```python
#CSV - comma separated values 
import csv

data = [
    ['student_ID', 'Name', 'Family name'],
    [1, 'Peter', 'FISer'],
    [2, 'Nastja', 'de la FIS']
]

#write csv
with open('students.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)
    
#read csv
with open('students.csv', 'r') as file:
    reader = csv.DictReader(file) # Loads each row as a dictionary
    for row in reader:
        print(f"Student {row['student_ID']} is {row['Name']} {row['Family name']}")
        
        
```

```python
#JSON (JavaScript Object Notation) is text-based format for storing and transporting data.
#alternative to XML

import json

# dictionary as input
target_profile = {
    "id": None,
    "name": "Sarah Connor",
    "isTerminated": False,
    "params": [0, 170, 60],
    "address": None
}

# convert from dict to JSON
json_string = json.dumps(target_profile, indent=1)
print(json_string)

# Save to a .json file
with open('target_profile.json', 'w') as file:
    json.dump(target_profile, file, indent=4)
```

```python
# Reading from a .json file
with open('target_profile.json', 'r') as file:
    data = json.load(file)

# Now 'data' is a regular Python dictionary again
print(f"{data['name']} was terminated: {data['isTerminated']}")
```
