````python
import pika

credentials = pika.PlainCredentials('martin', 'martin00')
parameters =  pika.ConnectionParameters('149.62.71.186', credentials=credentials)
connection = pika.BlockingConnection(parameters)
channel = connection.channel()

message = "Vsebina poslanega sporočila"

channel.basic_publish(exchange='my_new_topic', routing_key='small.fox.original', body=message)
channel.basic_publish(exchange='my_new_topic', routing_key='big.fox.original', body=message)
channel.basic_publish(exchange='my_new_topic', routing_key='small.cat.original', body=message)
channel.basic_publish(exchange='my_new_topic', routing_key='big.cat.original', body=message)


connection.close()
````


## 1. Authentication & Connection
The first few lines establish a secure link to the RabbitMQ server.
* **Credentials:** It logs in as user `martin`.
* **Connection:** It connects to the server at `149.62.71.186`.
* **Channel:** It opens a communication channel, which is where the commands are sent.

## 2. The Message Content
```python
message = "Vsebina poslanega sporočila"
```
The variable `message` contains the string "Content of the sent message" in Slovenian. This is the "payload" that will be delivered to the queues.

## 3. Publishing with "Dot" Routing Keys
The most important part of this script is how it publishes the messages:
```python
channel.basic_publish(exchange='my_new_topic', routing_key='small.fox.original', body=message)
```
Unlike the simple "info" or "error" keys from before, these keys use a **dot-separated format** (`speed.animal.type`). This is the signature style of a **Topic Exchange**.



### Why use these dots?
A Topic Exchange allows consumers to filter messages using **wildcards**:
* `*` (star) replaces exactly **one** word.
* `#` (hash) replaces **zero or more** words.

In your code, you are sending four distinct messages:
1.  `small.fox.original`
2.  `big.fox.original`
3.  `small.cat.original`
4.  `big.cat.original`

## 4. How a Consumer would "Catch" these
Because you used these specific keys, a receiver could choose exactly what to "subscribe" to:
* **To get everything about cats:** A consumer would listen for `*.cat.*`.
* **To get only the "big" animals:** A consumer would listen for `big.#`.
* **To get only "original" foxes:** A consumer would listen for `*.fox.original`.

## 5. Clean Exit
```python
connection.close()
```
Since this script is just a "fire and forget" producer, it sends the four messages and then immediately closes the connection to free up resources on the server.
