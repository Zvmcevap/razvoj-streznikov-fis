```python
import requests
import json
```

```python
res = requests.get('http://127.0.0.1:1234/getall')
```

```python
json.loads(res.text)
```

```python
res = requests.post('http://127.0.0.1:1234', json={'ime': 'Ziga', 'visina' : '180'})
print(res.text)
```

```python
res = requests.put('http://127.0.0.1:1234', json={'ime': 'Martin', 'visina' : '153'})
print(res.text)
```

```python
res = requests.delete('http://149.62.71.186:1234', json={'ime': 'Martin'})
print(res.text)
```
